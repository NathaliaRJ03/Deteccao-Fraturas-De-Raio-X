```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

# onde contem as imagens
pasta_caminho = '/content/runs/detect/predict-27'

# lista todos os arquivos da pasta
files = os.listdir(pasta_caminho)

# linha opcional
imagem_arquivos = [f for f in files if f.endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp', '.JPG'))]

#exibir as imagens / loop
for imagem_arq in imagem_arquivos:
  imagem_caminho = os.path.join(pasta_caminho, imagem_arq)
  imagem = mpimg.imread(imagem_caminho)
  plt.imshow(imagem)
  plt.axis('off')
  plt.show()
```
